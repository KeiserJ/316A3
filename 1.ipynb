```python
import pandas as pd

print("正在读取原始超大文件，请稍候...")
df = pd.read_csv('DATA/Gravity_V202211.csv')

# 1. 过滤年份：保留 1980 到 2020 年的数据
print("正在过滤 1980 - 2020 年的数据...")
df_filtered = df[(df['year'] >= 2000) & (df['year'] <= 2020)]

# 2. 极限瘦身法：只保留前端 3D 地球真正用到的 4 列数据！(这步能砍掉 80% 的体积)
print("正在剔除多余列，进行极限压缩...")
columns_to_keep = ['year', 'iso3num_o', 'iso3num_d', 'tradeflow_baci']
df_optimized = df_filtered[columns_to_keep]

# 3. 进一步压缩：剔除没有贸易额（空值或0）的无效行
df_optimized = df_optimized.dropna(subset=['tradeflow_baci'])
df_optimized = df_optimized[df_optimized['tradeflow_baci'] > 0]

# 4. 保存为供 GitHub Pages 专用的轻量化文件
output_filename = 'DATA/Gravity_2000_2020.csv'
df_optimized.to_csv(output_filename, index=False)

print(f"处理完成！文件已保存为: {output_filename}")
```
